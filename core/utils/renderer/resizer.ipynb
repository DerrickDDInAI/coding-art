```python
import numpy as np
import cv2
from get_resize_interpolation import get_interpolation
from resizer import resize_with_pad, resize_with_crop
```

```python
# read images
ref_img_path = "/Users/derrickvanfrausum/BeCode_AI/git-repos/coding-art/core/assets/images/mosaics/square_quadsections.png"
dest_img_path = "/Users/derrickvanfrausum/BeCode_AI/git-repos/learn-ai/opencv/assets/images/magritte/Magritte_TheSonOfMan.jpg"
ref_img = cv2.imread(ref_img_path)
dest_img = cv2.imread(dest_img_path)

# get images shapes
ref_img_height, ref_img_width, ref_img_channel = ref_img.shape
dest_img_height, dest_img_width, dest_img_channel = dest_img.shape

# print images shapes
print(f"ref_img: {(ref_img_height, ref_img_width, ref_img_channel)}, dest_img: {dest_img_height, dest_img_width, dest_img_channel}")

# get aspect ratio of both images
ref_img_ratio = ref_img.shape[1]/ref_img.shape[0]
dest_img_ratio = dest_img.shape[1]/dest_img.shape[0]
print(f"ref_img_ratio: {ref_img_ratio}, dest_img_ratio: {dest_img_ratio}")
```

```python
# if same aspect ratio, no need for padding
if dest_img_ratio == ref_img_ratio:
    out_img = dest_img

# else, padding required
else:

    # if dest_img_ratio is smaller than ref_img_ratio, add padding to width to have the same aspect ratio as reference
    if dest_img_ratio < ref_img_ratio:
        # get new width
        new_width = round(dest_img_height*ref_img_ratio)
        out_img = np.zeros((dest_img_height, new_width, dest_img_channel), dtype=dest_img.dtype)

        # get yx positions to center dest image
        dest_y, dest_x = 0, (new_width - dest_img_width)//2

    # if dest_img_ratio is smaller than ref_img_ratio, add padding to height to have the same aspect ratio as reference
    else:
        # get new height
        new_height = round(dest_img_width/ref_img_ratio)
        out_img = np.zeros((new_height, dest_img_width, dest_img_channel), dtype=dest_img.dtype)

        # get yx positions to center dest image
        dest_y, dest_x = (new_height - dest_img_height)//2, 0

    # paste dest image at the center of black canvas
    out_img[dest_y:dest_y+dest_img.shape[0], dest_x:dest_x+dest_img.shape[1]] = dest_img

# get interpolation
interpolation = get_interpolation(src_img_shape=dest_img.shape, out_img_shape=ref_img.shape)

# resize dest image with zeros padding (to get the reference ratio) to reference shape
out_resized_img = cv2.resize(out_img, (ref_img.shape[1], ref_img.shape[0]), interpolation=interpolation)

# show alongside reference image and dest image with padding and resized
cv2.imshow("ref",ref_img)
cv2.imshow("padding",out_img)

cv2.imshow("result",cv2.hconcat([ref_img, out_resized_img]))

# wait for any press on keyboard
cv2.waitKey(0)

# destroy all windows
cv2.destroyAllWindows()
cv2.waitKey(1) # workaround on mac to effectively close the windows
```

## crop to get same aspect ratio as reference image

```python
# if same aspect ratio, no need for cropping
if dest_img_ratio == ref_img_ratio:
    out_img = dest_img

# else, cropping required
else:

    # if dest_img_ratio is smaller than ref_img_ratio, crop width to have the same aspect ratio as reference
    if dest_img_ratio > ref_img_ratio:
        # get new width
        new_width = round(dest_img_height*ref_img_ratio)

        # get yx positions to center dest image
        dest_y, dest_x = 0, (dest_img_width - new_width)//2

    # if dest_img_ratio is bigger than ref_img_ratio, crop height to have the same aspect ratio as reference
    else:
        # get new height
        new_height = round(dest_img_width/ref_img_ratio)

        # get yx positions to center dest image
        dest_y, dest_x = (dest_img_height - new_height)//2, 0

        # paste dest image at the center of black canvas
        out_img = dest_img[dest_y:dest_y+new_height, dest_x:dest_x+dest_img.shape[1]]

# get interpolation
interpolation = get_interpolation(src_img_shape=dest_img.shape, out_img_shape=ref_img.shape)

# resize image (after cropping to get the reference ratio) to reference shape
out_resized_img = cv2.resize(out_img, (ref_img.shape[1], ref_img.shape[0]), interpolation=interpolation)

# show alongside reference image and dest image with padding and resized
cv2.imshow("ref",ref_img)
cv2.imshow("padding",out_img)

cv2.imshow("result",cv2.hconcat([ref_img, out_resized_img]))

# wait for any press on keyboard
cv2.waitKey(0)

# destroy all windows
cv2.destroyAllWindows()
cv2.waitKey(1) # workaround on mac to effectively close the windows
```
