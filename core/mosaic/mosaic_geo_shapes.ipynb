````python
def find_nearest_multiple(x, base):
    """
    Function to find the nearest multiple of base given an integer
    """
    return base * round(x/base)
````

````python
def create_mosaic_with_circles(dest_img, nb_rows, nb_cols, col_range=None, thickness=-1):
    """
    Function to recreate an image with circles

    Arguments
    * dest_img: destination image, i.e. image to recreate with circles
    """

    # get img height & width
    dest_img_height, dest_img_width = dest_img.shape[:2]

    # get best number of rows and cols to cover the whole img
    out_img_height = find_nearest_multiple(dest_img_height, nb_rows)
    out_img_width = find_nearest_multiple(dest_img_width, nb_cols)

    # resize image to recreate so that the grid can cover the whole image
    # get interpolation
    interpolation = get_interpolation(src_img_shape=(dest_img_height, dest_img_width), out_img_shape=(out_img_height, out_img_width))
    dest_img = cv2.resize(dest_img, (out_img_width, out_img_height), interpolation=interpolation)

    # get cell height & width
    cell_height, cell_width = out_img_height//nb_rows, out_img_width//nb_cols

    # get list of grid positions
    if col_range is not None:
        grid_positions = [(grid_y, grid_x) for grid_x in range(*col_range) for grid_y in range(nb_rows)]
        out_img = np.zeros_like(dest_img[:,col_range[0]:col_range[1]])
    else:
        grid_positions = [(grid_y, grid_x) for grid_x in range(nb_cols) for grid_y in range(nb_rows)]
        out_img = np.zeros_like(dest_img)

    # get index range of grid positions
    grid_positions_idxs = np.arange(len(grid_positions))
    # # choose random seed to recreate same shuffle or change it to see if you get better results
    # np.random.seed(1111)
    # shuffle grid positions (otherwise the first grids from top will get the best matching images)
    # np.random.shuffle(grid_positions_idxs)

    # iterate over each grid position
    for i in grid_positions_idxs:

        grid_y, grid_x = grid_positions[i]
        
        # get region of interest in img to recreate
        y = grid_y * cell_height
        x = grid_x * cell_width
        dest_roi = dest_img[y:y+cell_height, x:x+cell_width]
        
        # get mean of roi
        dest_roi_mean = cv2.mean(dest_roi)[:-1]

        # get center coordinates
        center_y, center_x = (y+cell_height//2), (x+cell_width//2)

        # Radius of circle
        # radius = min(cell_height//2, cell_width//2)
        min_cell_side = min(cell_height//2, cell_width//2)
        radius = np.random.randint(min_cell_side, min_cell_side*8)

        # min_cell_side = min(cell_height//2, cell_width//2)
        # max_cell_side = max(cell_height//2, cell_width//2)
        
        # # (major axis length, minor axis length)
        # axesLength = (max_cell_side, min_cell_side)
        
        # # angle = 0
        # angle = np.random.randint(0, 360)
        
        # startAngle = np.random.randint(0, 360)
        
        # endAngle = np.random.randint(startAngle, 360)

        # # draw ellipse
        # out_img = cv2.ellipse(out_img, (center_x, center_y), axesLength,
        #         angle, startAngle, endAngle, dest_roi_mean, thickness)

        # draw circle on roi
        out_img = cv2.circle(out_img, (center_x, center_y), radius, dest_roi_mean, thickness)

    return out_img
````

If third argument is False, you will get a polylines joining all the points, not a closed shape.
cv.polylines() can be used to draw multiple lines. Just create a list of all the lines you want to draw and pass it to the function. All lines will be drawn individually. It is a much better and faster way to draw a group of lines than calling cv.line() for each line.

````python
def create_mosaic_with_geo_shapes(
    dest_img, nb_rows, nb_cols, col_range=None, 
    geo_shape="circle", thickness=-1, line_type=cv2.LINE_AA,
    text = None,
    colormap=None,
    radius_factor=8,
    nb_polylines=1,
    nb_pts=3,
    max_font_scale=5
    ):
    """
    Function to recreate an image with geometric shapes

    Arguments
    * dest_img: destination image, i.e. image to recreate with geometric shapes
    """

    if text is not None:
        text_length = len(text)

        font = cv2.FONT_HERSHEY_SIMPLEX


    # get img height & width
    dest_img_height, dest_img_width = dest_img.shape[:2]

    # get best number of rows and cols to cover the whole img
    out_img_height = find_nearest_multiple(dest_img_height, nb_rows)
    out_img_width = find_nearest_multiple(dest_img_width, nb_cols)

    # resize image to recreate so that the grid can cover the whole image
    # get interpolation
    interpolation = get_interpolation(src_img_shape=(dest_img_height, dest_img_width), out_img_shape=(out_img_height, out_img_width))
    dest_img = cv2.resize(dest_img, (out_img_width, out_img_height), interpolation=interpolation)

    if colormap is not None:
        dest_img = cv2.cvtColor(dest_img, cv2.COLOR_BGR2GRAY)
        dest_img = cv2.applyColorMap(dest_img, colormap)

    # get cell height & width
    cell_height, cell_width = out_img_height//nb_rows, out_img_width//nb_cols

    # get list of grid positions
    if col_range is not None:
        grid_positions = [(grid_y, grid_x) for grid_x in range(*col_range) for grid_y in range(nb_rows)]
        out_img = np.zeros_like(dest_img[:,col_range[0]:col_range[1]])
    else:
        grid_positions = [(grid_y, grid_x) for grid_x in range(nb_cols) for grid_y in range(nb_rows)]
        out_img = np.zeros_like(dest_img)

    # get index range of grid positions
    grid_positions_idxs = np.arange(len(grid_positions))
    # # choose random seed to recreate same shuffle or change it to see if you get better results
    # np.random.seed(1111)
    # shuffle grid positions (otherwise the first grids from top will get the best matching images)
    # np.random.shuffle(grid_positions_idxs)

    # iterate over each grid position
    for i in grid_positions_idxs:

        grid_y, grid_x = grid_positions[i]
        
        # get region of interest in img to recreate
        y = grid_y * cell_height
        x = grid_x * cell_width
        dest_roi = dest_img[y:y+cell_height, x:x+cell_width]
        
        # get mean of roi
        dest_roi_mean = cv2.mean(dest_roi)[:-1]

        # get center coordinates
        center_y, center_x = (y+cell_height//2), (x+cell_width//2)

        if geo_shape == "circle":

            # Radius of circle
            min_cell_side = min(cell_height//2, cell_width//2)
            radius = np.random.randint(min_cell_side, min_cell_side*radius_factor)

            # draw circle on roi
            out_img = cv2.circle(out_img, (center_x, center_y), radius, dest_roi_mean, thickness)

        elif geo_shape == "ellipse":

            min_cell_side = min(cell_height//2, cell_width//2)
            max_cell_side = max(cell_height//2, cell_width//2)
        
            # (major axis length, minor axis length)
            axesLength = (np.random.randint(max_cell_side, max_cell_side*8), np.random.randint(min_cell_side, min_cell_side*8))
            
            # angle = 0
            angle = np.random.randint(0, 360)
            
            startAngle = np.random.randint(0, 360)
            
            endAngle = np.random.randint(startAngle, 360)
            
            # draw ellipse
            out_img = cv2.ellipse(out_img, (center_x, center_y), axesLength,
                    angle, startAngle, endAngle, dest_roi_mean, thickness)

        elif geo_shape == "rectangle":
            top_left = (np.random.randint(x, center_x), np.random.randint(y, center_y))
            bottom_right = (np.random.randint(center_x, center_x + (cell_width//2)), np.random.randint(center_y, center_y + (cell_height//2)))
            # top_left = (np.random.randint(center_x - (cell_width), center_x - cell_width//2), np.random.randint(center_y - (cell_height), center_y - cell_height//2))
            # bottom_right = (np.random.randint(center_x + cell_width//2, center_x + (cell_width)), np.random.randint(center_y + cell_height//2, center_y + (cell_height)))

            # draw rectangle
            out_img = cv2.rectangle(out_img, top_left, bottom_right, dest_roi_mean, thickness)

        elif geo_shape == "poly":
                pts_x = np.random.randint(x, x+cell_width, size=(nb_polylines*nb_pts, 1), dtype=np.int32)
                pts_y = np.random.randint(y, y+cell_height, size=(nb_polylines*nb_pts, 1), dtype=np.int32)
                pts = np.hstack((pts_x, pts_y)).reshape(nb_polylines,nb_pts, 2)
                                
                isClosed = True

                if thickness > 0:
                    # draw polylines
                    cv2.polylines(out_img, pts, isClosed, dest_roi_mean, thickness, lineType=cv2.LINE_AA)

                else:
                    # fill poly with colors
                    cv2.fillPoly(out_img, pts, dest_roi_mean, lineType=cv2.LINE_AA)
                

        elif geo_shape == "text" and text is not None:

            # get random text idx
            text_idx = np.random.randint(0,text_length)

            # get random text
            text_str = text[text_idx]

            # get random font scale
            font_scale = np.random.uniform(0,max_font_scale)

            # get boundary of this text
            text_size = cv2.getTextSize(text_str, font, font_scale, thickness)[0]

            # get coords based on boundary
            text_x = (center_x - text_size[0] // 2)
            text_y = (center_y + text_size[1] // 2)

            # add text
            out_img = cv2.putText(out_img, text_str, (text_x, text_y), font, font_scale,
            dest_roi_mean, thickness, line_type, False)

    return out_img
````

````python
def create_random_mosaic_with_geo_shapes(
    img=None,
    out_img_shape=None,
    nb_iterations=1,
    geo_shapes=["circle"], thickness=1, line_type=cv2.LINE_AA,
    text = None,
    colormap=None,
    radius_factor=2,
    nb_polylines=1,
    nb_pts=3,
    max_font_scale=5,
    font_thickness=1
    ):
    """
    Function to randomly draw geometric shapes

    Arguments
    * 
    """

    if text is not None:
        text_length = len(text)

        font = cv2.FONT_HERSHEY_SIMPLEX

    # get out img height & width
    if out_img_shape is None:
        out_img_shape = img.shape
    
    out_img_height, out_img_width = out_img_shape[:2]

    # create canvas if no image provided
    if img is None:
        out_img = np.zeros(out_img_shape, dtype=np.uint8)

    else:
        out_img = img

    photo = cv2.imread("/Users/derrickvanfrausum/BeCode_AI/git-repos/learn-ai/opencv/assets/images/avatar_dd/db/avatar_dd.png")

    for i in range(nb_iterations):

        previous_image = out_img.copy()
        mask = np.zeros(out_img_shape[:2], dtype = np.uint8)
        canvas = np.zeros(out_img_shape, dtype = np.uint8)

        for geo_shape in geo_shapes:

            if geo_shape == "circle":

                # get center coordinates
                center_y, center_x = np.random.randint(0, out_img_height), np.random.randint(0, out_img_width)

                # Radius of circle
                radius_min = 1
                radius_max = np.max((out_img_height//2, out_img_width//2))
                radius = np.random.randint(1, radius_max*radius_factor)

                # set random color
                color = np.random.randint(0,255, size=(3), dtype=np.uint8).tolist()

                # draw circle on roi
                # out_img = cv2.circle(out_img, (center_x, center_y), radius, color, thickness)
                mask = cv2.circle(mask, (center_x, center_y), radius, 1, thickness)

                # # add level of opacity according to size of shape
                # # inversely proportional to size
                # alpha = (radius - radius_min) / (radius_max - radius_min)
                # beta = 1 - alpha
                # out_img = cv2.addWeighted(out_img, alpha, previous_image, beta, 0)
                # out_img = cv2.addWeighted(out_img, alpha, previous_image, 0.5, 0)

                # # proportional to size
                # beta = (radius - radius_min) / (radius_max - radius_min)
                # alpha = 1 - beta
                # out_img = cv2.addWeighted(out_img, alpha, previous_image, beta, 0)

                # resize photo to bbox shape
                bbox_side = radius * 2
                bbox_shape = (bbox_side, bbox_side)
                bbox = (center_y - radius, center_y + radius, center_x - radius, center_x + radius,)

                try:
                    resized_photo = resize_with_crop(img = photo, ref_img_shape = (bbox_shape[0], bbox_shape[1], 3))

                    # put img to bbox position, clip img if bbox falls outside of canvas
                    clip_amount_min_y = 0 - bbox[0] if bbox[0] < 0 else 0
                    clip_amount_min_x = 0 - bbox[2] if bbox[2] < 0 else 0
                    clip_amount_max_y = out_img_height - bbox[1] if bbox[1] > out_img_height else 0
                    clip_amount_max_x = out_img_width - bbox[3] if bbox[3] > out_img_width else 0

                    canvas[
                        bbox[0] + clip_amount_min_y:bbox[1] + clip_amount_max_y, 
                        bbox[2] + clip_amount_min_x:bbox[3] + clip_amount_max_x, 
                        :] = resized_photo[
                        clip_amount_min_y:bbox_shape[0] + clip_amount_max_y, 
                        clip_amount_min_x:bbox_shape[1] + clip_amount_max_x
                        , :]
                    
                    masked_img = cv2.bitwise_and(canvas, canvas, mask=mask)

                except:
                    # img = np.ones((bbox_shape[0], bbox_shape[1], 3), dtype=np.uint8)
                    resized_photo = resize_with_crop(img = photo, ref_img_shape = out_img_shape)
                    masked_img = cv2.bitwise_and(resized_photo, resized_photo, mask=mask)

                    error_count += 1
                    print("error_count:", error_count)
                    print("bbox:", bbox)
                    print("bbox_shape:", bbox_shape)

            
            
                # out_img = cv2.add(out_img, img)









                size = np.pi * (radius ** 2)
                size_min = np.pi * (radius_min ** 2)
                size_max = np.pi * (radius_max ** 2)
                
                beta = (size - size_min) / (size_max - size_min)
                alpha = 1 - beta
                # out_img = cv2.addWeighted(out_img, alpha, previous_image, beta, 0)
                # out_img = cv2.addWeighted(out_img, beta, previous_image, alpha, 0)
                out_img = cv2.addWeighted(masked_img, beta, previous_image, alpha, 0)
                # out_img = cv2.addWeighted(masked_img, alpha, previous_image, beta, 0)


            elif geo_shape == "ellipse":
            
                # (major axis length, minor axis length)
                axesLength = (np.random.randint(1, out_img_height*radius_factor), np.random.randint(1, out_img_width*radius_factor))
                
                # angle = 0
                angle = np.random.randint(0, 360)
                
                startAngle = np.random.randint(0, 360)
                
                endAngle = np.random.randint(startAngle, 360)

                # set random color
                color = np.random.randint(0,255, size=(3), dtype=np.uint8).tolist()
                
                # draw ellipse
                out_img = cv2.ellipse(out_img, (center_x, center_y), axesLength,
                        angle, startAngle, endAngle, color, thickness)

                size = np.mean(axesLength)
                size_min = 1
                size_max = out_img_height*radius_factor

                beta = (size - size_min) / (size_max - size_min)
                alpha = 1 - beta
                # out_img = cv2.addWeighted(out_img, alpha, previous_image, beta, 0)
                out_img = cv2.addWeighted(out_img, beta, previous_image, alpha, 0)

            elif geo_shape == "rectangle":
                ys = np.random.randint(0, out_img_height, size=(2,))
                xs = np.random.randint(0, out_img_width, size=(2,))
                min_y = np.min(ys)
                min_x = np.min(xs)
                max_y = np.max(ys)
                max_x = np.max(xs)

                top_left = (min_x, min_y)
                bottom_right = (max_x, max_y)
                # top_left = (np.random.randint(center_x - (cell_width), center_x - cell_width//2), np.random.randint(center_y - (cell_height), center_y - cell_height//2))
                # bottom_right = (np.random.randint(center_x + cell_width//2, center_x + (cell_width)), np.random.randint(center_y + cell_height//2, center_y + (cell_height)))

                # set random color
                color = np.random.randint(0,255, size=(3), dtype=np.uint8).tolist()

                # draw rectangle
                out_img = cv2.rectangle(out_img, top_left, bottom_right, color, thickness)

                size = max_y * max_x
                size_min = 1
                size_max = out_img_height * out_img_width

                beta = (size - size_min) / (size_max - size_min)
                alpha = 1 - beta
                # out_img = cv2.addWeighted(out_img, alpha, previous_image, beta, 0)
                out_img = cv2.addWeighted(out_img, beta, previous_image, alpha, 0)


            elif geo_shape == "poly":
                pts_x = np.random.randint(0, out_img_width, size=(nb_polylines*nb_pts, 1), dtype=np.int32)
                pts_y = np.random.randint(0, out_img_height, size=(nb_polylines*nb_pts, 1), dtype=np.int32)
                pts = np.hstack((pts_x, pts_y)).reshape(nb_polylines,nb_pts, 2)
                                
                isClosed = True

                # set random color
                color = np.random.randint(0,255, size=(3), dtype=np.uint8).tolist()

                if thickness > 0:
                    # draw polylines
                    cv2.polylines(out_img, pts, isClosed, color, thickness, lineType=cv2.LINE_AA)

                else:
                    # fill poly with colors
                    cv2.fillPoly(out_img, pts, color, lineType=cv2.LINE_AA)


                poly_bbox = cv2.boundingRect(pts)
            
                size = (poly_bbox[2] - poly_bbox[0]) * (poly_bbox[3] - poly_bbox[1])
                size_min = 1
                size_max = out_img_height * out_img_width

                beta = (size - size_min) / (size_max - size_min)
                alpha = 1 - beta
                # out_img = cv2.addWeighted(out_img, alpha, previous_image, beta, 0)
                out_img = cv2.addWeighted(out_img, beta, previous_image, alpha, 0)
                    

            elif geo_shape == "text" and text is not None:

                # get random text idx
                text_idx = np.random.randint(0,text_length)

                # get random text
                text_str = text[text_idx]

                # get random font scale
                font_scale = np.random.uniform(0,max_font_scale)

                # get boundary of this text
                text_size = cv2.getTextSize(text_str, font, font_scale, thickness)[0]

                # get center coordinates
                center_y, center_x = np.random.randint(0, out_img_height), np.random.randint(0, out_img_width)

                # get coords based on boundary
                text_x = (center_x - text_size[0] // 2)
                text_y = (center_y + text_size[1] // 2)
        
                # set random color
                color = np.random.randint(0,255, size=(3), dtype=np.uint8).tolist()

                # add text
                out_img = cv2.putText(out_img, text_str, (text_x, text_y), font, font_scale,
                color, font_thickness, line_type, False)

                size = text_size[0] * text_size[1]
                size_min = 1
                size_max = out_img_height * out_img_width

                beta = (size - size_min) / (size_max - size_min)
                alpha = 1 - beta
                # out_img = cv2.addWeighted(out_img, alpha, previous_image, beta, 0)
                out_img = cv2.addWeighted(out_img, beta, previous_image, alpha, 0)

    if colormap is not None:
        out_img = cv2.cvtColor(out_img, cv2.COLOR_BGR2GRAY)
        out_img = cv2.applyColorMap(out_img, colormap)

    return out_img
````
