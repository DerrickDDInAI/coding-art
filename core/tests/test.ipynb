## bilateralFilter 
can reduce unwanted noise very well while keeping edges fairly sharp. However, it is very slow compared to most filters.

Sigma values: For simplicity, you can set the 2 sigma values to be the same. If they are small (< 10), the filter will not have much effect, whereas if they are large (> 150), they will have a very strong effect, making the image look "cartoonish".

Filter size: Large filters (d > 5) are very slow, so it is recommended to use d=5 for real-time applications, and perhaps d=9 for offline applications that need heavy noise filtering.

## Speed up video reading
Source: https://pyimagesearch.com/2017/02/06/faster-video-file-fps-with-cv2-videocapture-and-opencv/

```python
# import the necessary packages
from imutils.video import FileVideoStream
from imutils.video import FPS
import numpy as np
import argparse
import imutils
import time
import cv2
# # construct the argument parse and parse the arguments
# ap = argparse.ArgumentParser()
# ap.add_argument("-v", "--video", required=True,
# 	help="path to input video file")
# args = vars(ap.parse_args())

args = {"video": "/Users/derrickvanfrausum/Desktop/P1933689.MP4"}

# start the file video stream thread and allow the buffer to
# start to fill
print("[INFO] starting video file thread...")
fvs = FileVideoStream(args["video"]).start()
time.sleep(1.0)
# start the FPS timer
fps = FPS().start()

# loop over frames from the video file stream
while fvs.more():
	# grab the frame from the threaded video file stream, resize
	# it, and convert it to grayscale (while still retaining 3
	# channels)
	frame = fvs.read()
	if frame is None:
		break
	frame = imutils.resize(frame, width=450)
	frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
	frame = np.dstack([frame, frame, frame])
	# display the size of the queue on the frame
	cv2.putText(frame, "Queue Size: {}".format(fvs.Q.qsize()),
		(10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)	
	# show the frame and update the FPS counter
	cv2.imshow("Frame", frame)
	cv2.waitKey(1)
	fps.update()

# stop the timer and display FPS information
fps.stop()
print("[INFO] elasped time: {:.2f}".format(fps.elapsed()))
print("[INFO] approx. FPS: {:.2f}".format(fps.fps()))
# do a bit of cleanup
cv2.destroyAllWindows()
cv2.waitKey(1) # workaround to effectively close window on mac
fvs.stop()
```

```python
# Python3 code to find all three angles
# of a triangle given coordinate
# of all three vertices
import math

# returns square of distance b/w two points
def lengthSquare(X, Y):
	xDiff = X[0] - Y[0]
	yDiff = X[1] - Y[1]
	return xDiff * xDiff + yDiff * yDiff
	
def printAngle(A, B, C):
	
	# Square of lengths be a2, b2, c2
	a2 = lengthSquare(B, C)
	b2 = lengthSquare(A, C)
	c2 = lengthSquare(A, B)

	# length of sides be a, b, c
	a = math.sqrt(a2);
	b = math.sqrt(b2);
	c = math.sqrt(c2);

	# From Cosine law
	alpha = math.acos((b2 + c2 - a2) /
						(2 * b * c));
	betta = math.acos((a2 + c2 - b2) /
						(2 * a * c));
	gamma = math.acos((a2 + b2 - c2) /
						(2 * a * b));

	# Converting to degree
	alpha = alpha * 180 / math.pi;
	betta = betta * 180 / math.pi;
	gamma = gamma * 180 / math.pi;

	# printing all the angles
	print("alpha : %f" %(alpha))
	print("betta : %f" %(betta))
	print("gamma : %f" %(gamma))
		
# Driver code
A = (0, 0)
B = (5.90386,0)
C = (5.90386, 4.07625)

printAngle(A, B, C);

# This code is contributed
# by ApurvaRaj
```

```python
def normalize_between_range(xs, a, b):
    """
    Function to normalize a series between range [a,b]
    """
    min_x = np.min(xs)
    max_x = np.max(xs)
    return (b - a) * (xs - min_x)/(max_x - min_x) + a
```

```python
def construct_eq_triangle(centroid, radius):
    side_length = radius * np.sqrt(3)
    # Calculate three vertices of the triangle
    a = [centroid[0], centroid[1] + (np.sqrt(3) / 3) * side_length]  # Top vertex
    b = [centroid[0] - (side_length / 2), centroid[1] - (np.sqrt(3) / 6) * side_length]  # Bottom left vertex
    c = [centroid[0] + (side_length / 2), centroid[1] - (np.sqrt(3) / 6) * side_length]  # Bottom right vertex

    return a, b, c
```

```python
import numpy as np
from scipy.spatial.distance import euclidean
from fastdtw import fastdtw

def evaluate_similarity(movement1, movement2):
    """
    Evaluate the similarity of two body movements using the Dynamic Time Warping (DTW) algorithm
    :param movement1: array with the first body movement
    :param movement2: array with the second body movement
    :return: percentage of similarity
    """
    distance, path = fastdtw(movement1, movement2, dist=euclidean)
    similarity = 1 - (distance / np.sum(np.abs(movement1 - movement2)))
    return similarity
```

```python
import cv2
import numpy as np
from numba import jit

@jit(nopython=True)
def apply_gravity_forces(image):
    """
    Apply gravity forces based on color values.
    Darker pixels "attract" their neighbors more strongly, effectively pulling their colors towards them.
    """
    height, width, channels = image.shape
    new_image = np.zeros((height, width, channels), dtype=np.uint8)

    for x in range(1, height-1):
        for y in range(1, width-1):
            # Calculating the gravity force for the current pixel
            gravity_force = np.sum(image[x, y]) / (3 * 255)
            for dx in [-1, 0, 1]:
                for dy in [-1, 0, 1]:
                    if dx == 0 and dy == 0:
                        continue
                    neighbor_force = np.sum(image[x + dx, y + dy]) / (3 * 255)
                    force_diff = gravity_force - neighbor_force
                    # Apply the force difference to "pull" colors
                    new_image[x, y] = np.clip(image[x, y] + force_diff * 10, 0, 255).astype(np.uint8)

    return new_image

# Load the image
image_path = "/Users/derrickvanfrausum/Pictures/JWT_Carina_Nebula.png"
image = cv2.imread(image_path)

# Prepare video writer
fourcc = cv2.VideoWriter_fourcc(*'avc1')
out = cv2.VideoWriter('gravity_simulation_2.mp4', fourcc, 20.0, (image.shape[1], image.shape[0]))

iterations = 200  # Number of iterations for the simulation

# Apply gravity forces and write each iteration to the video
for _ in range(iterations):
    image = apply_gravity_forces(image)
    out.write(image)

out.release()

'Done'
```

```python
import cv2
import numpy as np

def interpolate_frames(frame1, frame2):
    # Convert frames to grayscale for optical flow calculation
    prev_gray = cv2.cvtColor(frame1, cv2.COLOR_BGR2GRAY)
    next_gray = cv2.cvtColor(frame2, cv2.COLOR_BGR2GRAY)

    # Calculate the optical flow between the two frames
    flow = cv2.calcOpticalFlowFarneback(prev_gray, next_gray, None, 0.5, 3, 15, 3, 5, 1.2, 0)

    # Create an interpolated frame
    height, width = frame1.shape[:2]
    flow_map = np.column_stack((np.repeat(np.arange(height), width), np.tile(np.arange(width), height)))
    displacement = flow_map + flow.reshape(-1, 2).astype(np.int16)
    displacement = np.clip(displacement, 0, [height-1, width-1])
    interpolated = frame2[displacement[:, 0], displacement[:, 1]].reshape(height, width, 3)

    return interpolated

# Open your video
cap = cv2.VideoCapture('/Users/derrickvanfrausum/DD_AI/slow-mo-fight-trim.mp4')

# Read the first frame
ret, prev_frame = cap.read()

# Loop over the video
while True:
    ret, next_frame = cap.read()
    if not ret:
        break

    for i in range(3):
    # Interpolate frames
    interpolated_frame = interpolate_frames(prev_frame, next_frame)

    # Here you can save or display the interpolated frame
    # For example, display the frame
    h,w,c = prev_frame.shape[0]//5, prev_frame.shape[1]//5, prev_frame.shape[2]//5
    cv2.imshow('Interpolated Frame', resize_with_crop(prev_frame, ref_img_shape=(h,w,c)))
    cv2.imshow('Interpolated Frame', resize_with_crop(interpolated_frame, ref_img_shape=(h,w,c)))
    cv2.imshow('Interpolated Frame', resize_with_crop(next_frame, ref_img_shape=(h,w,c))) 
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

    # Update the previous frame
    prev_frame = next_frame

# Release the video capture object
cap.release()
cv2.destroyAllWindows()
```

```python
def interpolate_frames(frame1, frame2, num_inter_frames):
    # Convert frames to grayscale for optical flow calculation
    gray1 = cv2.cvtColor(frame1, cv2.COLOR_BGR2GRAY)
    gray2 = cv2.cvtColor(frame2, cv2.COLOR_BGR2GRAY)

    # Calculate forward and backward optical flow
    forward_flow = cv2.calcOpticalFlowFarneback(gray1, gray2, None, 0.5, 3, 15, 3, 5, 1.2, 0)
    backward_flow = cv2.calcOpticalFlowFarneback(gray2, gray1, None, 0.5, 3, 15, 3, 5, 1.2, 0)

    height, width, channels = frame1.shape
    interpolated_frames = []

    for i in range(1, num_inter_frames + 1):
        fraction = i / (num_inter_frames + 1)
        
        # Map for forward interpolation
        forward_map_x, forward_map_y = np.meshgrid(np.arange(width), np.arange(height))
        forward_map_x = (forward_map_x + fraction * forward_flow[..., 0]).astype(np.float32)
        forward_map_y = (forward_map_y + fraction * forward_flow[..., 1]).astype(np.float32)
        
        # Map for backward interpolation
        backward_map_x, backward_map_y = np.meshgrid(np.arange(width), np.arange(height))
        backward_map_x = (backward_map_x - (1 - fraction) * backward_flow[..., 0]).astype(np.float32)
        backward_map_y = (backward_map_y - (1 - fraction) * backward_flow[..., 1]).astype(np.float32)

        # Remap the frames
        interp_frame_forward = cv2.remap(frame1, forward_map_x, forward_map_y, cv2.INTER_LINEAR)
        interp_frame_backward = cv2.remap(frame2, backward_map_x, backward_map_y, cv2.INTER_LINEAR)
        
        # Blend the frames
        interp_frame = cv2.addWeighted(interp_frame_forward, 1 - fraction, interp_frame_backward, fraction, 0)
        
        interpolated_frames.append(interp_frame)

    return interpolated_frames

# Video processing loop (omitted for brevity)

```
