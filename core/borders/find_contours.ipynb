## Find contours

```python
# convert the image to grayscale format
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# invert image
img_gray = cv2.bitwise_not(img_gray)

# apply binary thresholding
ret, thresh = cv2.threshold(img_gray, 150, 255, cv2.THRESH_BINARY) # 150, 255
# visualize the binary image
cv2.imshow('Binary image', thresh)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.waitKey(1)
```

```python
# detect the contours on the binary image using cv2.CHAIN_APPROX_NONE
contours, hierarchy = cv2.findContours(image=thresh, mode=cv2.RETR_TREE, method=cv2.CHAIN_APPROX_NONE)
                                     
# draw contours on the original image
img_copy = img.copy()
img_copy = cv2.drawContours(image=img_copy, contours=contours, contourIdx=-1, color=(255, 0, 0), thickness=5, lineType=cv2.LINE_AA)
               
# see the results
cv2.imshow('None approximation', img_copy)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.waitKey(1)
```

### try to center contour in middle of a square
dist = m - c
dist/2 = (m - nc)
dist/2 - m = - nc
m - dist/2 = nc
m - (m - c)/2 = nc

```python
thickness = -1

factor = 0.5
# get new contours centered in the middle
m = img.shape[0]//2
new_contours = [(m - (m - contour)/factor).astype(np.uint) for contour in contours]

# draw contours on the original image
img_copy = img.copy()
img_copy = cv2.drawContours(image=img_copy, contours=new_contours, contourIdx=-1, color=(255, 0, 0), thickness=thickness, lineType=cv2.LINE_AA)
               
# see the results
cv2.imshow('None approximation', img_copy)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.waitKey(1)
```
