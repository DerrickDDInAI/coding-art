## Read image

```python
img = cv2.imread("assets/images/pixel_transfers/db/picasso-the-weeping-woman.jpeg")
ref_img_height, ref_img_width = img.shape[:2]
print(ref_img_height, ref_img_width)
```

## Define variable & constants

```python
BORDER_COLOR = (0,0,0)
NB_IMGS = 60
# border_types = [cv2.BORDER_REFLECT, cv2.BORDER_REFLECT101, cv2.BORDER_REPLICATE, cv2.BORDER_CONSTANT, cv2.BORDER_ISOLATED, cv2.BORDER_TRANSPARENT, cv2.BORDER_WRAP, cv2.BORDER_DEFAULT]
border_types = [cv2.BORDER_REFLECT, cv2.BORDER_REFLECT101, cv2.BORDER_REPLICATE, cv2.BORDER_CONSTANT, cv2.BORDER_ISOLATED, cv2.BORDER_WRAP, cv2.BORDER_DEFAULT]
border_range_height = np.linspace(0, ref_img_height*20, NB_IMGS, dtype=np.uint16)
border_range_width = np.linspace(0, ref_img_height*20, NB_IMGS, dtype=np.uint16)
```

## add borders & save images

```python
out_img_dir = "constant"

for i, (height_border, width_border) in enumerate(zip(border_range_height, border_range_width)):

    dst = cv2.copyMakeBorder(img, top=height_border, bottom=height_border, left=width_border, right=width_border, borderType=cv2.BORDER_CONSTANT, value=BORDER_COLOR)
    dst = cv2.resize(dst, (ref_img_width, ref_img_height), interpolation=cv2.INTER_AREA)

    # set output image path
    out_img_path = str(project.out_img_dir_dict[out_img_dir] / f"{i:09d}.jpg")

    # save output image
    cv2.imwrite(out_img_path, dst)
```

```python
out_img_dir = "reflect_wrap"
border_types = [cv2.BORDER_REFLECT, cv2.BORDER_WRAP]

for i, (height_border, width_border) in enumerate(zip(border_range_height, border_range_width)):

    dst = cv2.copyMakeBorder(img, top=height_border, bottom=height_border, left=width_border, right=width_border, borderType=border_types[i%len(border_types)], value=BORDER_COLOR)
    dst = cv2.resize(dst, (ref_img_width, ref_img_height), interpolation=cv2.INTER_AREA)

    # set output image path
    out_img_path = str(project.out_img_dir_dict[out_img_dir] / f"{i:09d}.jpg")

    # save output image
    cv2.imwrite(out_img_path, dst)

    # # show alongside reference image and dest image with padding and resized
    # cv2.imshow("result",dst)

    # # wait for any press on keyboard
    # cv2.waitKey(0)

    # # destroy all windows
    # cv2.destroyAllWindows()
    # cv2.waitKey(1) # workaround on mac to effectively close the windows
```

```python
out_img_dir = "reflect_wrap_resize_first"
border_types = [cv2.BORDER_REFLECT, cv2.BORDER_WRAP]

for i, (height_border, width_border) in enumerate(zip(border_range_height, border_range_width)):
    print(ref_img_width-width_border//2, ref_img_height-height_border//2)
    dst = cv2.resize(img, (ref_img_width-width_border//2 , ref_img_height-height_border//2), interpolation=cv2.INTER_AREA)
    dst = cv2.copyMakeBorder(img, top=height_border, bottom=height_border, left=width_border, right=width_border, borderType=border_types[i%len(border_types)], value=BORDER_COLOR)

    # set output image path
    out_img_path = str(project.out_img_dir_dict[out_img_dir] / f"{i:09d}.jpg")

    # save output image
    cv2.imwrite(out_img_path, dst)
```

```python
out_img_dir = "reflect_wrap_dst"
border_types = [cv2.BORDER_WRAP]
NB_IMGS = 60
border_range_height = np.linspace(0, ref_img_height/3, NB_IMGS, dtype=np.uint16)
border_range_width = np.linspace(0, ref_img_height/3, NB_IMGS, dtype=np.uint16)
dst = img.copy()

for i, (height_border, width_border) in enumerate(zip(border_range_height, border_range_width)):

    dst = cv2.copyMakeBorder(dst, top=height_border, bottom=height_border, left=width_border, right=width_border, borderType=border_types[i%len(border_types)], value=BORDER_COLOR)
    dst = cv2.resize(dst, (ref_img_width, ref_img_height), interpolation=cv2.INTER_AREA)

    # set output image path
    out_img_path = str(project.out_img_dir_dict[out_img_dir] / f"{i:09d}.jpg")

    # save output image
    cv2.imwrite(out_img_path, dst)
```

```python
out_img_dir = "reflect_wrap_random"
border_types = [cv2.BORDER_REFLECT, cv2.BORDER_WRAP]

for i, (height_border, width_border) in enumerate(zip(border_range_height, border_range_width)):

    border_type_idx = np.random.randint(0,len(border_types))

    dst = cv2.copyMakeBorder(img, top=height_border, bottom=height_border, left=width_border, right=width_border, borderType=border_types[border_type_idx], value=BORDER_COLOR)
    dst = cv2.resize(dst, (ref_img_width, ref_img_height), interpolation=cv2.INTER_AREA)

    # set output image path
    out_img_path = str(project.out_img_dir_dict[out_img_dir] / f"{i:09d}.jpg")

    # save output image
    cv2.imwrite(out_img_path, dst)
```
