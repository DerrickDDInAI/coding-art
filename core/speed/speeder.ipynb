## loop video and speed up each turn 

```python
# get current date
today = date.today().strftime("%Y%m%d")

# set input & output video path
video_path = Path("assets/images/db/pink_moon_trim.mp4")
out_path = project.project_dir / f"{video_path.stem}_{today}.mp4"

# Initialize video stream
video_cap = cv2.VideoCapture(str(video_path))

# get video parameters
video_nb_frames = int(video_cap.get(cv2.CAP_PROP_FRAME_COUNT))
video_fps = video_cap.get(cv2.CAP_PROP_FPS)
video_width = int(video_cap.get(cv2.CAP_PROP_FRAME_WIDTH))
video_height = int(video_cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

print(f"number of frames = {video_nb_frames}")
print(f"fps = {video_fps}")
print(f"video width = {video_width}")
print(f"video height = {video_height}")

# set codec for output video
codec = "H264"

# set output shape
out_height, out_width, out_channel = 1920, 1080, 3

# create a videoWriter object
fourcc = cv2.VideoWriter_fourcc(*codec)
out_video = cv2.VideoWriter(filename=str(out_path), fourcc=fourcc, fps=video_fps, frameSize=(out_width, out_height))

# get number of videos
# nb_videos = int(video_nb_frames//video_fps)

speed_factors = [i**2 for i in range(0, round(np.sqrt(video_nb_frames)), 2)]
speed_factors[0] = 1
# speed_factors = np.geomspace(video_nb_frames - 1, 1, num=video_nb_frames//30, dtype=np.uint)

# get number of high speed videos to add 5 seconds of high speed video (only 3 frames)
nb_high_speed_videos = round(2 * video_fps//3)
high_speed_factors = [video_nb_frames for _ in range(nb_high_speed_videos)]
speed_factors.extend(high_speed_factors)

# # set frame to keep when speed is too much that we need to select 3 frames to keep
# frame_to_keep_idx = 8 * video_fps


for speed_factor in speed_factors:
# for video_nb in range(1, nb_videos):
    # if speed_factor < 5:
    #     speed_factor = 5
    num = video_nb_frames//speed_factor
    if num < 5:
        num = 5

    frame_idxs = np.linspace(0, video_nb_frames - 1, num=num, dtype=np.uint)
    # if len(frame_idxs) < 3:
    #     frame_idxs = [0, frame_to_keep_idx, video_nb_frames - 1]
    # frame_idxs = np.linspace(0, video_nb_frames - 1, num=video_nb_frames//video_nb, dtype=np.uint)

    # Initialize video stream
    video_cap = cv2.VideoCapture(str(video_path))

    # iterate over frame indexes
    for frame_idx in frame_idxs:
        # set frame position to the index
        video_cap.set(cv2.CAP_PROP_POS_FRAMES, frame_idx)

        # read video stream
        ret, frame = video_cap.read()

        if not ret:
            print(f"frame {frame_idx} is empty")
            break

        # rotate & resize frame
        frame = np.rot90(frame, -1)
        frame = resize_with_crop(frame, ref_img_shape=(out_height, out_width, out_channel))

        # write output frame
        out_video.write(frame)

# release video stream & video rendering
video_cap.release()
out_video.release()
```

```python
# get current date
today = date.today().strftime("%Y%m%d")

# set input & output video path
video_path = Path("assets/images/db/pink_moon_trim.mp4")
out_path = project.project_dir / f"{video_path.stem}_{today}.mp4"

# Initialize video stream
video_cap = cv2.VideoCapture(str(video_path))

# get video parameters
video_nb_frames = int(video_cap.get(cv2.CAP_PROP_FRAME_COUNT))
video_fps = video_cap.get(cv2.CAP_PROP_FPS)
video_width = int(video_cap.get(cv2.CAP_PROP_FRAME_WIDTH))
video_height = int(video_cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

print(f"number of frames = {video_nb_frames}")
print(f"fps = {video_fps}")
print(f"video width = {video_width}")
print(f"video height = {video_height}")

# set codec for output video
codec = "H264"

# create a videoWriter object
fourcc = cv2.VideoWriter_fourcc(*codec)
out_video = cv2.VideoWriter(filename=str(out_path), fourcc=fourcc, fps=video_fps, frameSize=(video_width, video_height))

# get number of videos
# nb_videos = int(video_nb_frames//video_fps)

speed_factors = [i**2 for i in range(0, round(np.sqrt(video_nb_frames)), 2)]
speed_factors[0] = 1
# speed_factors = np.geomspace(video_nb_frames - 1, 1, num=video_nb_frames//30, dtype=np.uint)

# get number of high speed videos to add 5 seconds of high speed video (only 3 frames)
nb_high_speed_videos = round(2 * video_fps//3)
high_speed_factors = [video_nb_frames for _ in range(nb_high_speed_videos)]
speed_factors.extend(high_speed_factors)

# # set frame to keep when speed is too much that we need to select 3 frames to keep
# frame_to_keep_idx = 8 * video_fps

# set past_offset, to get past frames
past_offset = 4 * video_fps

for speed_factor in speed_factors:
# for video_nb in range(1, nb_videos):
    # if speed_factor < 5:
    #     speed_factor = 5
    num = video_nb_frames//speed_factor
    if num < 4:
        num = 4

    frame_idxs = np.linspace(0, video_nb_frames - 1, num=num, dtype=np.uint)
    # if len(frame_idxs) < 3:
    #     frame_idxs = [0, frame_to_keep_idx, video_nb_frames - 1]
    # frame_idxs = np.linspace(0, video_nb_frames - 1, num=video_nb_frames//video_nb, dtype=np.uint)

    # Initialize video stream
    video_cap = cv2.VideoCapture(str(video_path))

    # iterate over frame indexes
    for frame_idx in frame_idxs:
        # set frame position to the index
        video_cap.set(cv2.CAP_PROP_POS_FRAMES, frame_idx)

        # read video stream
        ret, frame = video_cap.read()

        if not ret:
            print(f"frame {frame_idx} is empty")
            break

        # set a past frame position (x second earlier)
        video_cap.set(cv2.CAP_PROP_POS_FRAMES, frame_idx - past_offset)

        # read video stream
        past_ret, past_frame = video_cap.read()

        if not past_ret:
            past_frame = np.zeros_like(frame)

        # write output frame
        out_video.write(cv2.add(frame, past_frame))

# release video stream & video rendering
video_cap.release()
out_video.release()
```
