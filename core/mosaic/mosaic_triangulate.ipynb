## The idea

Rebuild an image out of triangles instead of square grid cells:

1. Pick feature points across the source image — `cv2.goodFeaturesToTrack`
   finds corners, or sample more densely wherever the local variance is high,
   so that detailed regions get smaller triangles.
2. Connect them with `scipy.spatial.Delaunay`, which maximises the smallest
   angle in the mesh and therefore avoids long thin slivers.
3. Fill each triangle with the mean colour of the source pixels inside it
   (`cv2.fillPoly` with a mask, or `cv2.drawContours`).
